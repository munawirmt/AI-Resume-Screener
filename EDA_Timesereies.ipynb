```python
##Time sampling
df_tesla.head()
```

```python
df_tesla.set_index('Date', inplace=True)
df_tesla.resample('A').mean()
```

```python
df_tesla.resample('A').min()
```

```python
df_tesla.resample('A').max()
```

```python
df_tesla.resample('A').max()['Open'].plot(kind='bar')
```

```python
## quarterly start frequency
df_tesla.resample('BA').max()
```

```python
df_tesla.resample('BQS').mean()
```

```python
## PLOTTING
df_tesla['Open'].resample('M').max().plot(kind='bar',figsize=(15,6))
```

```python
df_tesla['High'].rolling(11).max().head(20)  ## rolling
```

```python
df_tesla['Open : 30 days rolling']=df_tesla['Open'].rolling(30).mean()
df_tesla.head(31)
```

```python
df_tesla[['Open','Open : 30 days rolling']].plot(figsize=(15,6))
```
